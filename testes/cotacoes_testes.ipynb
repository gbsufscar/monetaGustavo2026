```python
import sys
import os

# Adiciona o diretório raiz ao path para importar os módulos
sys.path.insert(0, os.path.abspath('..'))

import yfinance as yf
from datetime import datetime, timedelta
import pandas as pd
from simbolos import obter_tickers_filtrados, obter_classificacoes_besst

# Obtém os símbolos de ações do código principal
# Opção 1: Todos os tickers brasileiros
todos_tickers = obter_tickers_filtrados()
simbolos_todos = list(todos_tickers.keys())

print(f"📊 Total de tickers disponíveis: {len(simbolos_todos)}")
print(f"Primeiros 10: {simbolos_todos[:10]}\n")

# Opção 2: Filtrar por classificação BESST (ex: apenas "Energia")
tickers_escolhidos = obter_tickers_filtrados(classificacoes_besst=['Energia'])
simbolos_escolhidos = list(tickers_escolhidos.keys())

print(f"📈 Tickers 'Escolhida': {len(simbolos_escolhidos)}")
print(f"Exemplos: {simbolos_escolhidos[:5]}\n")

# Opção 3: Usar apenas alguns tickers para teste
simbolos = simbolos_escolhidos[:5]  # Pega os primeiros 5 tickers da classificação "Escolhida"

print(f"🎯 Testando com: {simbolos}\n")

# Download das cotações
data_inicio = (datetime.now() - timedelta(days=365)).strftime('%Y-%m-%d')
data_fim = datetime.now().strftime('%Y-%m-%d')

cotacoes: pd.DataFrame = yf.download(simbolos, start=data_inicio, end=data_fim)['Adj Close']

print("✅ Cotações obtidas:")
print(cotacoes.head())
print(f"\nShape: {cotacoes.shape}")
```
